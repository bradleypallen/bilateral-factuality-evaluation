```python
from factuality_assessor import FactualityAssessor
from prompts import UNILATERAL_PROMPT_CONTRASTIVE, BILATERAL_PROMPT_CONTRASTIVE
from IPython.display import display, Markdown
```

```python
MODEL = "claude-3-5-sonnet-20241022"
```

```python
EXAMPLES = [
    {
        "problem": "Who was the first man to walk on the Moon?",
        "answer": "Neil Armstrong"
    }, 
    # {
    #     "problem": "Who was the first man to walk on the Moon?",
    #     "answer": "Pete Conrad"
    # }, 
    # {
    #     "problem": "Are penguins birds?",
    #     "answer": "No"
    # },
    # {
    #     "problem": "Who was the frobnoz that restified the branks?",
    #     "answer": "Rutherford B. Hayes"
    # },
    {
        "problem": "When was America discovered?",
        "answer": "1492"
    },
    {
        "problem": "Did Lee Harvey Oswald act alone in the assassination of John Kennedy?",
        "answer": "Yes"
    },
    # {
    #     "problem": "Is the Holy Trinity one entity or three entities?",
    #     "answer": "Three"
    # },
    {
        "problem": "When 49 g of KClO3 decomposes, the resulting O2 reacts with 10.8 g of impure metal (20% purity) to form metal oxide. Calculate the amount of carbon needed to convert the metal oxide back to pure metal. The metal is amphoteric in nature and is one of the most abundant metals in earth crust.",
        "answer": "0.72 g"
    },
    {
        "problem": "What's the name of the women's liberal arts college in Cambridge, Massachusetts?",
        "answer": "Radcliffe College"
    }
]
```

```python
def unilateral_evaluation(example):
    unilateral_evaluator = FactualityAssessor(MODEL, UNILATERAL_PROMPT_CONTRASTIVE)
    evaluation = unilateral_evaluator.invoke(example)
    display(Markdown(f"## Unilateral"))
    display(Markdown(f"### Reasoning"))
    display(Markdown(evaluation["reasoning"][0]))
    display(Markdown(f"### Evaluation: {evaluation['evaluation']}"))
```

```python
def bilateral_evaluation(example):
    bilateral_evaluator = FactualityAssessor(MODEL, BILATERAL_PROMPT_CONTRASTIVE)
    evaluation = bilateral_evaluator.invoke(example)
    display(Markdown(f"## Bilateral"))
    display(Markdown(f"### Reasoning"))
    display(Markdown(evaluation["reasoning"][0]))
    display(Markdown(f"### Evaluation: {evaluation['evaluation']}"))
```

```python
for example in EXAMPLES:
    display(Markdown(f'# {example["problem"]} {example["answer"]}'))
    unilateral_evaluation(example)
    bilateral_evaluation(example)
```
