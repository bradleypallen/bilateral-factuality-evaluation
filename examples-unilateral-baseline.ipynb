```python
from judges import UnilateralJudge
from IPython.display import display, Markdown
```

```python
MODEL = "google/gemini-2.0-flash-001"
```

```python
EXAMPLES = [
    {   
        "question": "Chocolate is made from material",
        "answer": "Sugar"
    },
    {
        "question": "Who was the first man to walk on the Moon?",
        "answer": "Neil Armstrong"
    }, 
    {
        "question": "Who was the first man to walk on the Moon?",
        "answer": "Pete Conrad"
    }, 
    {
        "question": "Are penguins birds?",
        "answer": "No"
    },
    {
        "question": "Who was the frobnoz that restified the branks?",
        "answer": "Rutherford B. Hayes"
    },
    {
        "question": "When was America discovered?",
        "answer": "1492"
    },
    {
        "question": "Did Lee Harvey Oswald act alone in the assassination of John Kennedy?",
        "answer": "Yes"
    },
    {
        "question": "Is the Holy Trinity one entity or three entities?",
        "answer": "Three"
    },
]
```

```python
def unilateral_evaluation(example):
    unilateral_evaluator = UnilateralJudge(MODEL, prompt_type='few')
    evaluation = unilateral_evaluator.invoke("test_examples", example, samples=3)
    display(Markdown(f"### Reasoning"))
    display(Markdown(f"#### Evaluations"))
    for i, v in enumerate(evaluation["evaluations"]):
        display(Markdown(f"##### Evaluation {i+1}"))
        display(Markdown(f"{v}"))
    display(Markdown(f"### Evaluation: {evaluation['wk_v']}"))
    display(Markdown(f"### Execution time: {evaluation['execution_time']:.2f} seconds"))
    display(Markdown(f"### Tokens used: {evaluation['tokens_used']:5d}"))
```

```python
for example in EXAMPLES:
    display(Markdown(f'# {example["question"]} {example["answer"]}'))
    unilateral_evaluation(example)
```
