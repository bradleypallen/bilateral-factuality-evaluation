```python
def wk_truth_value(tv):
    if tv is None:
        return "e"
    elif tv:
        return "t"
    else:
        return "f"
```

```python
def bilateral_truth_value(tv):
    return f'<{wk_truth_value(tv[0])}, {wk_truth_value(tv[1])}>'
```

```python
def dot_exists(X):
    """
    Implementation of the restricted Kleene existential quantifier function (∃˙)
    as described in Definition 3 of the paper.
    
    Parameters:
    X -- A list of pairs [u, v], where each u and v is either True, None, or False
         (representing t, e, and f in the weak Kleene three-valued logic)
    
    Returns:
    A truth value True, None, or False (representing t, e, or f)
    
    According to Definition 3:
    ∃˙(X) = t if ⟨t, t⟩ ∈ X
            e if for all ⟨u, v⟩, either u = e or v = e
            f if ⟨t, t⟩ ∉ X and for some ⟨u, v⟩ ∈ X, u ≠ e and v ≠ e
    """
    # Check if <t, t> is in X
    for pair in X:
        if pair == [True, True]:
            return True
    
    # Check if all pairs contain at least one 'e' (None)
    all_contain_e = True
    for pair in X:
        if pair[0] is not None and pair[1] is not None:
            all_contain_e = False
            break
    
    if all_contain_e:
        return None
    
    # If we reach here, <t, t> is not in X and there's at least one pair
    # where neither element is 'e'
    return False


# Example usage
if __name__ == "__main__":
    # Case 1: <t, t> is in X
    X1 = [[True, True], [False, True], [None, None]]
    print(f"∃˙({X1}) = {dot_exists(X1)}")  # Should be True (t)
    
    # Case 2: All pairs have at least one 'e'
    X2 = [[None, True], [False, None], [None, None]]
    print(f"∃˙({X2}) = {dot_exists(X2)}")  # Should be None (e)
    
    # Case 3: <t, t> is not in X and some pair has no 'e'
    X3 = [[True, False], [False, True], [None, None]]
    print(f"∃˙({X3}) = {dot_exists(X3)}")  # Should be False (f)
```

```python
def dot_forall(X):
    """
    Implementation of the restricted Kleene universal quantifier function (∀˙)
    as described in Definition 3 of the paper.
    
    Parameters:
    X -- A list of pairs [u, v], where each u and v is either True, None, or False
         (representing t, e, and f in the weak Kleene three-valued logic)
    
    Returns:
    A truth value True, None, or False (representing t, e, or f)
    
    According to Definition 3:
    ∀˙(X) = t if ⟨t, f⟩, ⟨t, e⟩ ∉ X and for some ⟨u, v⟩ ∈ X, u ≠ e and v ≠ e
            e if for all ⟨u, v⟩ ∈ X, either u = e or v = e
            f if {⟨t, f⟩, ⟨t, e⟩} ∩ X ≠ ∅ and for some ⟨u, v⟩ ∈ X, either u = e or v = e
    """
    # Check if all pairs contain at least one 'e' (None)
    all_contain_e = True
    for pair in X:
        if pair[0] is not None and pair[1] is not None:
            all_contain_e = False
            break
    
    if all_contain_e:
        return None
    
    # Check if ⟨t, f⟩ or ⟨t, e⟩ is in X
    has_problematic_pair = False
    for pair in X:
        if (pair[0] is True and (pair[1] is False or pair[1] is None)):
            has_problematic_pair = True
            break
    
    # Check if there's at least one pair where at least one element is 'e'
    has_e_pair = False
    for pair in X:
        if pair[0] is None or pair[1] is None:
            has_e_pair = True
            break
    
    # Case 3: ⟨t, f⟩ or ⟨t, e⟩ is in X and there's a pair with at least one 'e'
    if has_problematic_pair and has_e_pair:
        return False
    
    # Case 1: ⟨t, f⟩ and ⟨t, e⟩ are not in X and there's at least one pair with no 'e'
    if not has_problematic_pair and not all_contain_e:
        return True
    
    # Default case (should not reach here if the definition is comprehensive)
    # But we'll return False as a fallback
    return False


# Example usage
if __name__ == "__main__":
    # Case 1: Neither <t, f> nor <t, e> is in X, and there's a pair with no 'e'
    X1 = [[False, True], [True, True], [None, None]]
    print(f"∀˙({X1}) = {dot_forall(X1)}")  # Should be True (t)
    
    # Case 2: All pairs have at least one 'e'
    X2 = [[None, True], [False, None], [None, None]]
    print(f"∀˙({X2}) = {dot_forall(X2)}")  # Should be None (e)
    
    # Case 3: <t, f> is in X and there's a pair with at least one 'e'
    X3 = [[True, False], [False, None], [None, True]]
    print(f"∀˙({X3}) = {dot_forall(X3)}")  # Should be False (f)
    
    # Case 3 (alt): <t, e> is in X and there's a pair with at least one 'e'
    X4 = [[True, None], [False, True], [None, None]]
    print(f"∀˙({X4}) = {dot_forall(X4)}")  # Should be False (f)
```

```python
penguins_are_birds = [True, False]
sparrows_are_birds = [True, False]
eagles_are_birds = [True, False]
penguins_fly = [False, True]
sparrows_fly = [True, False]
eagles_fly = [True, False]
```

```python
X0 = [[penguins_are_birds[0], penguins_fly[0]], [sparrows_are_birds[0], sparrows_fly[0]], [eagles_are_birds[0], eagles_fly[0]]]
print([ bilateral_truth_value(x) for x in X0 ])
X1 = [[penguins_are_birds[0], penguins_fly[1]], [sparrows_are_birds[0], sparrows_fly[1]], [eagles_are_birds[0], eagles_fly[1]]]
print([ bilateral_truth_value(x) for x in X1 ])
print(bilateral_truth_value([dot_forall(X0), dot_exists(X1)]))
```
